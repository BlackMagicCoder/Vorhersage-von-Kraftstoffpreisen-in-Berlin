# Vorhersage von Kraftstoffpreisen in Berlin
Data Mining, WiSe 2026/27

Ziel: Median-Preis (E10, Diesel) pro Tankstelle und Stunde, 24 Stunden im Voraus (Regression).

## 0. Setup und Daten laden
- Imports, Laden der Trainingsdaten, Bereinigung

## 1. Deskriptive Analyse (nur Trainingsdaten)
- Aufbau, Datentypen, Wertebereiche, fehlende und falsche Werte
- Tankstellen und Marken
- Tages- und Wochenverlauf, Entwicklung über die Zeit, Autokorrelation
- Was lernen wir daraus?

## 2. Merkmale
- Aus den Daten: frühere Preise (ab 24 h zurück), Stunde, Wochentag, Tankstelle/Marke
- Extern (Ideen): Rohölpreis, Feiertage/Ferien, Wetter
- Regel: nur Infos, die 24 h vorher bekannt waren; ohne Anpassung in Folgejahren nutzbar

## 3. Gütemaße
- Genauigkeit: MAE, RMSE
- Monetär: Ersparnis beim Tanken zur vorhergesagten günstigsten Stunde

## 4. Benchmark
- Preis wie vor 24 Stunden (nur erklären, Umsetzung in Teil 2)
